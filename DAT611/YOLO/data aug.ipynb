```python
import albumentations as A
import cv2
import os
import shutil
from pathlib import Path

transform = A.Compose([
    A.Rotate(limit=30, p=0.8),
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.3),
    A.GaussianBlur(p=0.2),
], bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels']))

images_dir = "dataset/images"
labels_dir = "dataset/labels"

for img_file in Path(images_dir).glob("*.jpg"):
    image = cv2.imread(str(img_file))
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    label_file = Path(labels_dir) / (img_file.stem + ".txt")
    bboxes, class_labels = [], []

    with open(label_file) as f:
        for line in f:
            cls, x, y, w, h = map(float, line.strip().split())
            bboxes.append([x, y, w, h])
            class_labels.append(int(cls))

    for i in range(3):  # create 3 augmented copies per image
        augmented = transform(image=image, bboxes=bboxes, class_labels=class_labels)

        out_img = cv2.cvtColor(augmented['image'], cv2.COLOR_RGB2BGR)
        out_name = f"{img_file.stem}_aug{i}"

        cv2.imwrite(f"{images_dir}/{out_name}.jpg", out_img)

        with open(f"{labels_dir}/{out_name}.txt", "w") as f:
            for bbox, cls in zip(augmented['bboxes'], augmented['class_labels']):
                f.write(f"{cls} {' '.join(map(str, bbox))}\n")

print("Augmentation complete!")
```
