```python

# Language: python

import os
import random
import shutil
import yaml
import csv
from ultralytics import YOLO

# Define dataset directories
base_dir = r"dataset"
images_dir = os.path.join(base_dir, "images")
labels_dir = os.path.join(base_dir, "labels")

# Define destination train/val directories
dest_dirs = {
    "train_images": os.path.join(base_dir, "train", "images"),
    "train_labels": os.path.join(base_dir, "train", "labels"),
    "val_images": os.path.join(base_dir, "val", "images"),
    "val_labels": os.path.join(base_dir, "val", "labels")
}

# Create destination directories if they don't exist
for d in dest_dirs.values():
    os.makedirs(d, exist_ok=True)

# Get list of all image files and shuffle
all_images = [f for f in os.listdir(images_dir) if f.lower().endswith(".jpg") or f.lower().endswith(".jpeg")]
random.shuffle(all_images)

# Calculate split index
split_index = int(0.8 * len(all_images))
train_images = all_images[:split_index]
val_images = all_images[split_index:]

def copy_files(file_list, src_img_dir, src_lbl_dir, dest_img_dir, dest_lbl_dir):
    for img_file in file_list:
        # Copy image file
        src_img_path = os.path.join(src_img_dir, img_file)
        dest_img_path = os.path.join(dest_img_dir, img_file)
        shutil.copy(src_img_path, dest_img_path)
        
        # Determine corresponding label file name (assuming same base name with .txt extension)
        base_name, _ = os.path.splitext(img_file)
        label_file = base_name + ".txt"
        src_lbl_path = os.path.join(src_lbl_dir, label_file)
        dest_lbl_path = os.path.join(dest_lbl_dir, label_file)
        if os.path.exists(src_lbl_path):
            shutil.copy(src_lbl_path, dest_lbl_path)

# Copy training and validation files
copy_files(train_images, images_dir, labels_dir, dest_dirs["train_images"], dest_dirs["train_labels"])
copy_files(val_images, images_dir, labels_dir, dest_dirs["val_images"], dest_dirs["val_labels"])

# Create YAML file for YOLO
yaml_data = {
    "path": base_dir,  # base directory
    "train": os.path.join("train", "images"),
    "val": os.path.join("val", "images"),
    "nc": 5,
    "names": ["50 Naira", "100 Naira", "200 Naira", "500 Naira", "1000 Naira"]
}
yaml_file_path = os.path.join(base_dir, "naira_dataset.yaml")
with open(yaml_file_path, "w") as file:
    yaml.dump(yaml_data, file)

```

# YoloV8

```python

# Load the largest pretrained YOLOv8 model
model = YOLO("yolov8x.pt")

# Train the model (remove invalid 'save_csv' flag)
train_results = model.train(data=yaml_file_path, epochs=200, save=True, save_txt=True)

# Validate the model
val_results = model.val(data=yaml_file_path)

# Save training metrics to CSV if available
csv_output_path = os.path.join(base_dir, "training_metrics.csv")
if hasattr(train_results, "metrics") and isinstance(train_results.metrics, dict):
    with open(csv_output_path, mode="w", newline="") as csv_file:
        writer = csv.writer(csv_file)
        writer.writerow(["Metric", "Value"])
        for key, value in train_results.metrics.items():
            writer.writerow([key, value])

# Advanced visualization: display training results if a plot method exists
if hasattr(train_results, "plot"):
    train_results.plot()
```
